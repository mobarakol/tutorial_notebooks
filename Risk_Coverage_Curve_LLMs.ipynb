```python
import numpy as np
import matplotlib.pyplot as plt

# Convert x values (exclusion ratio) to percentage
refer = np.array([0, 20, 40, 60, 80, 90])  # Now in percentage

# Convert y values (ROUGE-L scores) to percentage
avg_met_by_refer = np.array([35.7, 37.3, 39.0, 40.6, 41.4, 41.8])  # Vector-MoLoRA
avg_met_by_refer_2 = np.array([27.2, 30.4, 33.5, 36.7, 40.8, 43.0])  # LoRA
avg_met_by_refer_3 = np.array([33.7, 34.7, 35.7, 36.7, 37.4, 37.8])  # MoRA

plt.figure(figsize=(8, 6))
ax = plt.gca()

# Plot the data with markers
ax.plot(refer, avg_met_by_refer, marker='o', markersize=5, label='Llama')
ax.plot(refer, avg_met_by_refer_2, marker='s', markersize=5, label='MedGemma')
ax.plot(refer, avg_met_by_refer_3, marker='^', markersize=5, label='Qwen')

# Format x-axis tick labels with "%" sign
ax.set_xticks(refer)
ax.set_xticklabels([f"{int(x)}%" for x in refer])  # Convert each tick label to percentage format

# Labels and title
ax.legend(fontsize=15)
ax.set_ylabel('Performance (METEOR)', fontsize=15)
ax.set_xlabel("Refer to Clinicians\n(Exclusion Ratio of Highly Uncertain Samples)", fontsize=15)
ax.set_title('Performance Rejection Curve', fontsize=16)

# Format tick labels
ax.tick_params(axis='both', labelsize=14)

# Display grid for better readability
ax.grid(True, linestyle='--', alpha=0.7)

plt.show()
# Save the figure
plt.savefig("33-Risk_coverage_curve.pdf", bbox_inches='tight', dpi=1500)
plt.close()
```
