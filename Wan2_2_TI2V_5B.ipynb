```python
import os
import subprocess
from pathlib import Path
from huggingface_hub import login

from huggingface_hub import login

login(token="your toekn")

# 2) Clone repo if needed
if not Path("Wan2.2").exists():
    subprocess.run(["git", "clone", "https://github.com/Wan-Video/Wan2.2.git"], check=True)

os.chdir("Wan2.2")

# 3) Install requirements
# subprocess.run(["pip", "install", "-U", "pip", "setuptools", "wheel"], check=True)
# subprocess.run(["pip", "install", "-r", "requirements.txt"], check=True)

# 4) Download model if needed
model_dir = Path("./Wan2.2-TI2V-5B")
if not model_dir.exists():
    subprocess.run([
        "huggingface-cli", "download",
        "Wan-AI/Wan2.2-TI2V-5B",
        "--local-dir", str(model_dir)
    ], check=True)

# 5) Generate
prompt = "bike rides in a city, cinematic street documentary, realistic motion, crowded roads, colorful lights"

subprocess.run([
    "python", "generate.py",
    "--task", "ti2v-5B",
    "--size", "1280*704",
    "--ckpt_dir", str(model_dir),
    "--offload_model", "True",
    "--convert_model_dtype",
    "--t5_cpu",
    "--prompt", prompt,
], check=True)

print("Done")
```
